# LCEL: LangChain Expression Language

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
load_dotenv()

groq_api_key = os.environ["GROQ_API_KEY"]

model = ChatGroq(model="Gemma2-9b-It", groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.output_parsers import StrOutputParser

messages = [
    SystemMessage(content="Translate the following from English to French. Don't give any additional information"),
    HumanMessage(content="Hello how are you?")
]

response = model.invoke(messages)
response
```

```python
parser = StrOutputParser()
parser.invoke(response)
```

```python
from langchain.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","Translate the following into: {language}. Don't give additional text."),
        ("user", "{text}")
    ]
)
prompt.invoke({"language":"French","text":"Hello?"})
```

```python
# LCEL
chain = prompt|model|parser
chain.invoke({"language":"French","text":"Hello?"})
```
